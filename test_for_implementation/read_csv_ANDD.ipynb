```python
import pandas as pd
```

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
```

```python
df = pd.read_csv('scFv_from_ANDD_dataset_wu_et_al.csv', sep=';')
df
```

```python
list(df.columns)
```

```python
df[~(df['Ab/Nano_CDR H1'] == "\\") & ~(df['Ab/Nano_CDR L1'] == "\\")][:3]
```

```python
df_to_save = df[~(df['Ab/Nano_CDR H1'] == "\\") & ~(df['Ab/Nano_CDR L1'] == "\\")].reset_index(drop=True)
df_to_save
```

```python
df_to_save[:2]
```

```python
'X'*3
```

```python
df_to_save['heavy_with_X'] = df_to_save.apply(lambda x: x['Ab/Nano H_Chain AA'].replace(x['Ab/Nano_CDR H1'], 'X'*len(x['Ab/Nano_CDR H1'])).replace(x['Ab/Nano_CDR H2'], 'X'*len(x['Ab/Nano_CDR H2'])).replace(x['Ab/Nano_CDR H3'], 'X'*len(x['Ab/Nano_CDR H3'])), axis=1)
df_to_save['light_with_X'] = df_to_save.apply(lambda x: x['Ab/Nano L_Chain AA'].replace(x['Ab/Nano_CDR L1'], 'X'*len(x['Ab/Nano_CDR L1'])).replace(x['Ab/Nano_CDR L2'], 'X'*len(x['Ab/Nano_CDR L2'])).replace(x['Ab/Nano_CDR L3'], 'X'*len(x['Ab/Nano_CDR L3'])), axis=1)
df_to_save
```

```python
len([i for i in df_to_save['heavy_with_X'].iloc[0].split('X') if not i == ''])
```

```python
df_to_save['is_ok'] = df_to_save.apply(lambda x: len([i for i in x['heavy_with_X'].split('X') if not i == '']) == 4 and len([i for i in x['light_with_X'].split('X') if not i == '']) == 4, axis=1)
df_to_save['is_ok'].value_counts()
```

```python
df_to_save['id'] = range(1, len(df_to_save) + 1)
df_to_save['id'] = df_to_save.apply(lambda x: f"{x['PDB_ID']}-{x['id']}_{len(x['Ab/Nano H_Chain AA']) + 1}_{len(x['Ab/Nano H_Chain AA']) + 1 + 14}", axis=1)
df_to_save
```

```python
linker = 'GGGGS'*3 # Use a G4S linker repeated 3 times
df_to_save['seq_with_linker'] = df_to_save.apply(lambda x: f"{x['heavy_with_X']}{linker}{x['light_with_X']}", axis=1)
df_to_save
```

```python
df_to_save.head(2)
```

```python
with open('scFv_templates.fasta', 'w') as f:
    for _, row in df_to_save.iterrows():
        f.write(f">{row['id']}\n{row['seq_with_linker']}\n")
```
