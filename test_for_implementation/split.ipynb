```python
def identify_G4S_linker(scFv_chain):
    """
    Identify G4S linkers in a structure.

    Parameters
    ----------
    scFv_chain : biotite.structure.AtomArray
        The scFv chain to analyze.

    Returns
    -------
    list of tuples
        A list of tuples, where each tuple contains the indices of the atoms that form a G4S linker.
    """
    # scFv sequence
    scFv_seq = scFv_chain[scFv_chain.atom_name == 'CA'].res_name

    # Define the G4S sequence
    g4s_sequence = ['GLY', 'GLY', 'GLY', 'GLY', 'SER']
    
    # Initialize an empty list to store the indices of the G4S linkers
    g4s_linkers = []
    
    # Iterate over the residues in the structure
    for i in range(len(scFv_seq) - len(g4s_sequence) + 1):
        # Check if the current sequence matches the G4S sequence
        if list(scFv_seq[i:i + len(g4s_sequence)]) == g4s_sequence:
            # If it matches, store the indices of the atoms in the linker
            linker_indices = np.arange(i + 1, i + len(g4s_sequence) + 1)  # +1 to account for the fact that residue indices start at 1
            g4s_linkers.append(linker_indices)
    
    return g4s_linkers
```

```python
identify_G4S_linker(s[s.chain_id == 'B'])
```

```python
linker_indices = identify_G4S_linker(s)
linker_s = s[np.isin(np.arange(len(s)), np.concatenate(linker_indices))]
linker_s
```

```python
chain_H_mask = (s.chain_id == 'B') & (s.res_id <= 119)
chain_L_mask = (s.chain_id == 'B') & (s.res_id >= 135)
s.chain_id[chain_H_mask] = "H"
s.chain_id[chain_L_mask] = "L"
s_A_H_L = s[np.isin(s.chain_id, ['A', 'H', 'L'])]
print(len(s)), print(len(s_A_H_L))
s_A_H_L
```

```python
# Then, renumber each chain's residue IDs to start from 1
for chain in np.unique(s_A_H_L.chain_id):
    chain_mask = s_A_H_L.chain_id == chain
    s_A_H_L.res_id[chain_mask] = s_A_H_L.res_id[chain_mask] - np.min(s_A_H_L.res_id[chain_mask]) + 1
s_A_H_L
```

```python
save_structure("test_X_residue_no_ncaa_model_0_split.pdb", s_A_H_L)
```
